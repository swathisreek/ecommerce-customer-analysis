```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv("../data/ecommerce_customer_data.csv")

# Display first 5 rows
print("First 5 Customers:")
print(df.head())

# Dataset information
print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())

# Statistical summary
print("\nStatistical Summary:")
print(df.describe())

# Basic categorical analysis
print("\nCustomer Data Summary:")
for column in df.select_dtypes(include="object").columns:
    print(f"\n{column}:")
    print(df[column].value_counts())

# Correlation analysis
numeric_df = df.select_dtypes(include="number")

print("\nCorrelation Matrix:")
print(numeric_df.corr())

# Correlation heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("Customer Data Correlation Heatmap")
plt.show()

# Histograms for numerical columns
numeric_df.hist(figsize=(12, 8))
plt.suptitle("Distribution of Numerical Customer Data")
plt.tight_layout()
plt.show()

print("\nE-commerce Customer Analysis Completed Successfully!")
```
